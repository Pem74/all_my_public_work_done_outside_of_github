```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.ticker import MaxNLocator
```

```python
X = [1974, 1988, 1989]
Avoine = [1.5, 0.7, 0.5]
Ble = [19, 30, 32]
Mais = [7.5, 13, 12]
Orge = [9.5, 10.5, 10]
```

```python
plt.figure()
plt.xlabel("Années")
plt.ylabel("Production (en milliard de tonnes)")
plt.title("Les céréales cultivées en France")

plt.plot(X[0:2], Avoine[0:2], label='Avoine', linestyle='--', color=mcolors.CSS4_COLORS['yellowgreen'], marker='x')
plt.plot(X[1:3], Avoine[1:3], label='Avoine', linestyle='-', color=mcolors.CSS4_COLORS['yellowgreen'], marker='x')
plt.text(X[-1] + 0.5, Avoine[-1] - 0.5, 'Avoine', color=mcolors.CSS4_COLORS['yellowgreen'], fontweight='bold')
plt.plot(X[0:2], Ble[0:2], label='Blé', linestyle='--', color=mcolors.CSS4_COLORS['gold'], marker='x')
plt.plot(X[1:3], Ble[1:3], label='Blé', linestyle='-', color=mcolors.CSS4_COLORS['gold'], marker='x')
plt.text(X[-1] + 0.5, Ble[-1] - 0.5, 'Blé', color=mcolors.CSS4_COLORS['gold'], fontweight='bold')
plt.plot(X[0:2], Mais[0:2], label='Maïs', linestyle='--', color=mcolors.CSS4_COLORS['forestgreen'], marker='x')
plt.plot(X[1:3], Mais[1:3], label='Maïs', linestyle='-', color=mcolors.CSS4_COLORS['forestgreen'], marker='x')
plt.text(X[-1] + 0.5, Mais[-1] - 0.5, 'Maïs', color=mcolors.CSS4_COLORS['forestgreen'], fontweight='bold')
plt.plot(X[0:2], Orge[0:2], label='Orge', linestyle='--', color=mcolors.CSS4_COLORS['peru'], marker='x')
plt.plot(X[1:3], Orge[1:3], label='Orge', linestyle='-', color=mcolors.CSS4_COLORS['peru'], marker='x')
plt.text(X[-1] + 0.5, Orge[-1] - 0.5, 'Orge', color=mcolors.CSS4_COLORS['peru'], fontweight='bold')

plt.xlim(1973, 1992)
plt.gca().xaxis.set_major_locator(MaxNLocator(integer=True))
plt.grid()
plt.show()
```
